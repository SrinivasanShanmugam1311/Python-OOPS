```python
# ============================================================
# OOPS-02
# ASSIGNMENT 1 — LIST OF STUDENT OBJECTS
#              + FILTER + MAP + REDUCE
# ============================================================
#
# PURPOSE:
#
# Understand how OOPS objects can be stored inside a LIST
# and then processed using:
#
#       filter()
#       map()
#       reduce()
#
#
# ------------------------------------------------------------
# BASIC IDEA
# ------------------------------------------------------------
#
# Earlier we may have data like this:
#
#       ["Alex", 40, "M"]
#       ["Anu", 30, "F"]
#       ["Ravi", 25, "M"]
#
# Each record contains:
#
#       name
#       age
#       gender
#
#
# Instead of keeping them only as normal lists,
# create a Student OBJECT for every record.
#
#
# ------------------------------------------------------------
# CONVERSION
# ------------------------------------------------------------
#
# Normal record:
#
#       ["Alex", 40, "M"]
#
#               ↓
#
#       Student("Alex", 40, "M")
#
#               ↓
#
#         Student Object
#
#
# Do this for EVERY record.
#
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
# Original Records
#
#   ["Alex", 40, "M"]
#   ["Anu",  30, "F"]
#   ["Ravi", 25, "M"]
#
#           │
#           │ create objects
#           ↓
#
#       Student Objects
#
#       student1
#       student2
#       student3
#
#           │
#           │ append()
#           ↓
#
#       students
#
#       [
#          Student object,
#          Student object,
#          Student object
#       ]
#
#
# ------------------------------------------------------------
# IMPORTANT IDEA
# ------------------------------------------------------------
#
# The list is NOT storing only names or ages.
#
# It is storing references to Student OBJECTS.
#
#
#       students[0]
#            │
#            ↓
#       Student object
#            │
#       ┌────┼─────┐
#       ↓    ↓     ↓
#     name  age  gender
#
#
# Therefore:
#
#       students[0].name
#
# means:
#
#       first object
#           ↓
#       Student object
#           ↓
#          name
#
#
# ------------------------------------------------------------
# AFTER CREATING OBJECTS
# ------------------------------------------------------------
#
# We can PROCESS those objects:
#
# FILTER
# ------
# Select only objects that satisfy a condition.
#
# Example:
#
#       Keep only male students.
#
#
# MAP
# ---
# Change something for every object.
#
# Example:
#
#       Increase every student's age by 1.
#
#
# REDUCE
# ------
# Combine many object values into ONE result.
#
# Example:
#
#       Find total age.
#
#
# ------------------------------------------------------------
# EASY MEMORY RULE
# ------------------------------------------------------------
#
#       FILTER  → SELECT
#
#       MAP     → CHANGE
#
#       REDUCE  → COMBINE
#
#
# ============================================================


from functools import reduce


# ============================================================
# STUDENT CLASS
# ============================================================

class Student:

    def __init__(self, name, age, gender):

        # Store one student's information.
        self.name = name
        self.age = age
        self.gender = gender


    def display(self):

        # Display this Student object's data.
        print(
            self.name,
            self.age,
            self.gender
        )



# ============================================================
# STEP 1 — ORIGINAL DATA
# ============================================================

student_records = [

    ["Alex", 40, "M"],

    ["Anu", 30, "F"],

    ["Ravi", 25, "M"],

    ["Priya", 28, "F"]

]



# ============================================================
# STEP 2 — CREATE A LIST OF STUDENT OBJECTS
# ============================================================

students = []


for record in student_records:

    # record[0] = name
    # record[1] = age
    # record[2] = gender

    student = Student(
        record[0],
        record[1],
        record[2]
    )


    # IMPORTANT:
    #
    # We are appending the Student OBJECT.
    #
    # We are NOT appending only the student's name.

    students.append(student)



# ============================================================
# STEP 3 — DISPLAY ALL STUDENT OBJECTS
# ============================================================

print("ALL STUDENTS")
print("------------")


for student in students:

    student.display()



# ============================================================
# STEP 4 — FILTER
#
# FILTER = SELECT
#
# Select only male students.
# ============================================================

male_students = list(

    filter(

        lambda student:
            student.gender == "M",

        students
    )
)


print("\nMALE STUDENTS")
print("-------------")


for student in male_students:

    student.display()



# ============================================================
# STEP 5 — REDUCE
#
# REDUCE = COMBINE
#
# Combine all student ages into ONE total.
# ============================================================

total_age = reduce(

    lambda total, student:
        total + student.age,

    students,

    0
)


print("\nTOTAL AGE")
print("---------")

print(total_age)



# ============================================================
# STEP 6 — MAP
#
# MAP = CHANGE
#
# Increase every student's age by 1.
# ============================================================

def increase_age(student):

    student.age = student.age + 1

    return student


updated_students = list(

    map(
        increase_age,
        students
    )
)


print("\nAFTER INCREASING AGE BY 1")
print("-------------------------")


for student in updated_students:

    student.display()
```

```python
# ============================================================
# ASSIGNMENT 2 — __str__()
# ============================================================
#
# PURPOSE:
#
# Understand how Python decides what to display when
# we print an OBJECT.
#
#
# ------------------------------------------------------------
# WITHOUT __str__()
# ------------------------------------------------------------
#
# student1 = Student("Alex", 40)
#
# print(student1)
#
# Python may display something like:
#
# <__main__.Student object at 0x....>
#
# This is not very useful to us.
#
#
# ------------------------------------------------------------
# WITH __str__()
# ------------------------------------------------------------
#
# We can tell Python:
#
# "When somebody prints my object,
#  display it in THIS format."
#
#
#                     Student
#                        │
#                 ┌──────┴──────┐
#                 ↓             ↓
#               name           age
#                 │
#                 └──────┬──────┘
#                        ↓
#                     __str__()
#                        ↓
#                "Alex, 40"
#
#
# ------------------------------------------------------------
# EASY MEMORY RULE
# ------------------------------------------------------------
#
# __init__() → How should the OBJECT be CREATED?
#
# __str__()  → How should the OBJECT be PRINTED?
#
# ============================================================


class StudentString:

    def __init__(self, name, age):

        self.name = name
        self.age = age


    def __str__(self):

        # This string is returned whenever
        # we print this object.

        return f"{self.name}, {self.age}"


# ============================================================
# TEST
# ============================================================

student1 = StudentString(
    "Alex",
    40
)


print(student1)
```

```python
# ============================================================
# ASSIGNMENT 3 — STATIC METHOD
# ============================================================
#
# PURPOSE:
#
# Understand a method that belongs logically to a CLASS
# but does NOT need an object's data.
#
#
# ------------------------------------------------------------
# NORMAL INSTANCE METHOD
# ------------------------------------------------------------
#
# def display(self):
#
#     Uses "self"
#
#     ↓
#
#     Needs an OBJECT.
#
#
# ------------------------------------------------------------
# STATIC METHOD
# ------------------------------------------------------------
#
# @staticmethod
# def add(a, b):
#
#     No "self"
#
#     ↓
#
#     Does NOT need an object's data.
#
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
#                     MathUtility
#                         │
#              ┌──────────┼──────────┐
#              ↓          ↓          ↓
#             add()     square()    power()
#
#
# These functions are logically related to mathematics.
#
# So we keep them together under ONE class.
#
# But they do NOT need:
#
#       self
#
# because they are not working with one particular object.
#
#
# ------------------------------------------------------------
# EASY MEMORY RULE
# ------------------------------------------------------------
#
# Instance method:
#
#       object.method()
#       uses self
#
#
# Static method:
#
#       Class.method()
#       does NOT need self
#
# ============================================================


class MathUtility:

    @staticmethod
    def add(number1, number2):

        return number1 + number2


    @staticmethod
    def square(number):

        return number * number


    @staticmethod
    def power(number, exponent):

        return number ** exponent


# ============================================================
# TEST
# ============================================================

print(
    "Addition :",
    MathUtility.add(10, 20)
)


print(
    "Square :",
    MathUtility.square(5)
)


print(
    "Power :",
    MathUtility.power(2, 3)
)
```

```python
# ============================================================
# ASSIGNMENT 4 — CLASS METHOD
# ============================================================
#
# PURPOSE:
#
# Understand a method that works with the CLASS itself.
#
#
# ------------------------------------------------------------
# IMPORTANT DIFFERENCE
# ------------------------------------------------------------
#
# Normal method:
#
#       def method(self):
#
#       self → OBJECT
#
#
# Class method:
#
#       @classmethod
#       def method(cls):
#
#       cls → CLASS
#
#
# ------------------------------------------------------------
# EXAMPLE
# ------------------------------------------------------------
#
# Suppose data comes as ONE string:
#
#       "Alex,40,M"
#
# But Student needs:
#
#       name
#       age
#       gender
#
#
# We can use a class method to:
#
#       receive string
#           ↓
#       split string
#           ↓
#       get individual values
#           ↓
#       create Student object
#
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
#             "Alex,40,M"
#                  │
#                  ↓
#          from_string()
#                  │
#                split
#                  ↓
#          Alex   40   M
#                  │
#                  ↓
#              cls(...)
#                  │
#                  ↓
#          Student OBJECT
#
#
# ------------------------------------------------------------
# EASY MEMORY RULE
# ------------------------------------------------------------
#
# self → current OBJECT
#
# cls  → current CLASS
#
# ============================================================


class StudentFromString:

    def __init__(self, name, age, gender):

        self.name = name
        self.age = age
        self.gender = gender


    @classmethod
    def from_string(cls, student_data):

        # Split:
        #
        # "Alex,40,M"
        #
        # into:
        #
        # ["Alex", "40", "M"]

        values = student_data.split(",")


        name = values[0]
        age = int(values[1])
        gender = values[2]


        # cls means StudentFromString here.
        #
        # Therefore this creates a new object.

        return cls(
            name,
            age,
            gender
        )


    def display(self):

        print("Name   :", self.name)
        print("Age    :", self.age)
        print("Gender :", self.gender)


# ============================================================
# TEST
# ============================================================

student1 = StudentFromString.from_string(
    "Alex,40,M"
)


student1.display()
```

```python
# ============================================================
# ASSIGNMENT 5 — CLASS VARIABLE
# ============================================================
#
# PURPOSE:
#
# Understand the difference between:
#
#       OBJECT / INSTANCE VARIABLE
#
# and
#
#       CLASS VARIABLE
#
#
# ------------------------------------------------------------
# INSTANCE VARIABLE
# ------------------------------------------------------------
#
# Every student has their OWN:
#
#       name
#       age
#
#
# student1.name = "Alex"
#
# student2.name = "Anu"
#
#
# Different objects can have different values.
#
#
# ------------------------------------------------------------
# CLASS VARIABLE
# ------------------------------------------------------------
#
# Some information is COMMON to all students.
#
# Example:
#
#       college_name
#
#
# Instead of storing the same college name separately
# inside every object, we can store it at CLASS level.
#
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
#                    StudentCollege
#
#             college_name = "ABC College"
#                     SHARED
#                       │
#          ┌────────────┼────────────┐
#          ↓            ↓            ↓
#      student1      student2      student3
#        Alex           Anu          Ravi
#
#
# All objects can use the SAME class variable.
#
#
# ------------------------------------------------------------
# OBJECT COUNT
# ------------------------------------------------------------
#
# Another useful class variable:
#
#       student_count
#
# Every time we create an object:
#
#       student_count += 1
#
#
# ------------------------------------------------------------
# EASY MEMORY RULE
# ------------------------------------------------------------
#
# self.variable
#
#       ↓
#
# belongs to an OBJECT
#
#
# ClassName.variable
#
#       ↓
#
# belongs to the CLASS / shared by objects
#
# ============================================================


class StudentCollege:

    # CLASS VARIABLES
    #
    # Shared by all StudentCollege objects.

    college_name = "ABC College"

    student_count = 0


    def __init__(self, name, age):

        # INSTANCE VARIABLES
        #
        # Different for every object.

        self.name = name
        self.age = age


        # One new Student object has been created.

        StudentCollege.student_count += 1


    def display(self):

        print("Name    :", self.name)

        print("Age     :", self.age)

        print(
            "College :",
            StudentCollege.college_name
        )


# ============================================================
# CREATE OBJECTS
# ============================================================

student1 = StudentCollege(
    "Alex",
    40
)


student2 = StudentCollege(
    "Anu",
    30
)


student3 = StudentCollege(
    "Ravi",
    25
)


# ============================================================
# TEST
# ============================================================

student1.display()

print()

student2.display()

print()

student3.display()


print(
    "\nTotal Students :",
    StudentCollege.student_count
)
```
