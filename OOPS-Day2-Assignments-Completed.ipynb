```python
# ============================================================
# OOPS-02
# ASSIGNMENT 1 — LIST OF STUDENT OBJECTS
#              + FILTER + MAP + REDUCE
# ============================================================
#
# PURPOSE:
#
# Understand how OOPS objects can be stored inside a LIST
# and then processed using:
#
#       filter()
#       map()
#       reduce()
#
#
# ------------------------------------------------------------
# BASIC IDEA
# ------------------------------------------------------------
#
# Earlier we may have data like this:
#
#       ["Alex", 40, "M"]
#       ["Anu", 30, "F"]
#       ["Ravi", 25, "M"]
#
# Each record contains:
#
#       name
#       age
#       gender
#
#
# Instead of keeping them only as normal lists,
# create a Student OBJECT for every record.
#
#
# ------------------------------------------------------------
# CONVERSION
# ------------------------------------------------------------
#
# Normal record:
#
#       ["Alex", 40, "M"]
#
#               ↓
#
#       Student("Alex", 40, "M")
#
#               ↓
#
#         Student Object
#
#
# Do this for EVERY record.
#
#
# ------------------------------------------------------------
# MENTAL MODEL
# ------------------------------------------------------------
#
# Original Records
#
#   ["Alex", 40, "M"]
#   ["Anu",  30, "F"]
#   ["Ravi", 25, "M"]
#
#           │
#           │ create objects
#           ↓
#
#       Student Objects
#
#       student1
#       student2
#       student3
#
#           │
#           │ append()
#           ↓
#
#       students
#
#       [
#          Student object,
#          Student object,
#          Student object
#       ]
#
#
# ------------------------------------------------------------
# IMPORTANT IDEA
# ------------------------------------------------------------
#
# The list is NOT storing only names or ages.
#
# It is storing references to Student OBJECTS.
#
#
#       students[0]
#            │
#            ↓
#       Student object
#            │
#       ┌────┼─────┐
#       ↓    ↓     ↓
#     name  age  gender
#
#
# Therefore:
#
#       students[0].name
#
# means:
#
#       first object
#           ↓
#       Student object
#           ↓
#          name
#
#
# ------------------------------------------------------------
# AFTER CREATING OBJECTS
# ------------------------------------------------------------
#
# We can PROCESS those objects:
#
# FILTER
# ------
# Select only objects that satisfy a condition.
#
# Example:
#
#       Keep only male students.
#
#
# MAP
# ---
# Change something for every object.
#
# Example:
#
#       Increase every student's age by 1.
#
#
# REDUCE
# ------
# Combine many object values into ONE result.
#
# Example:
#
#       Find total age.
#
#
# ------------------------------------------------------------
# EASY MEMORY RULE
# ------------------------------------------------------------
#
#       FILTER  → SELECT
#
#       MAP     → CHANGE
#
#       REDUCE  → COMBINE
#
#
# ============================================================


from functools import reduce


# ============================================================
# STUDENT CLASS
# ============================================================

class Student:

    def __init__(self, name, age, gender):

        # Store one student's information.
        self.name = name
        self.age = age
        self.gender = gender


    def display(self):

        # Display this Student object's data.
        print(
            self.name,
            self.age,
            self.gender
        )



# ============================================================
# STEP 1 — ORIGINAL DATA
# ============================================================

student_records = [

    ["Alex", 40, "M"],

    ["Anu", 30, "F"],

    ["Ravi", 25, "M"],

    ["Priya", 28, "F"]

]



# ============================================================
# STEP 2 — CREATE A LIST OF STUDENT OBJECTS
# ============================================================

students = []


for record in student_records:

    # record[0] = name
    # record[1] = age
    # record[2] = gender

    student = Student(
        record[0],
        record[1],
        record[2]
    )


    # IMPORTANT:
    #
    # We are appending the Student OBJECT.
    #
    # We are NOT appending only the student's name.

    students.append(student)



# ============================================================
# STEP 3 — DISPLAY ALL STUDENT OBJECTS
# ============================================================

print("ALL STUDENTS")
print("------------")


for student in students:

    student.display()



# ============================================================
# STEP 4 — FILTER
#
# FILTER = SELECT
#
# Select only male students.
# ============================================================

male_students = list(

    filter(

        lambda student:
            student.gender == "M",

        students
    )
)


print("\nMALE STUDENTS")
print("-------------")


for student in male_students:

    student.display()



# ============================================================
# STEP 5 — REDUCE
#
# REDUCE = COMBINE
#
# Combine all student ages into ONE total.
# ============================================================

total_age = reduce(

    lambda total, student:
        total + student.age,

    students,

    0
)


print("\nTOTAL AGE")
print("---------")

print(total_age)



# ============================================================
# STEP 6 — MAP
#
# MAP = CHANGE
#
# Increase every student's age by 1.
# ============================================================

def increase_age(student):

    student.age = student.age + 1

    return student


updated_students = list(

    map(
        increase_age,
        students
    )
)


print("\nAFTER INCREASING AGE BY 1")
print("-------------------------")


for student in updated_students:

    student.display()
```
